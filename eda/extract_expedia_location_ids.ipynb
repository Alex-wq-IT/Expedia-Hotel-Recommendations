# Extract location/channel IDs from Expedia test.csv

Минимальная выгрузка ID-полей для последующего маппинга на реальные названия.

```python
import pandas as pd
from pathlib import Path
```

```python
DATA_PATH = Path('data/test.csv')
OUT_PATH = Path('test_location_ids.csv')
```

```python
cols = [
    'posa_continent',
    'user_location_country',
    'user_location_region',
    'user_location_city',
    'channel',
    'hotel_continent',
    'hotel_country',
    'hotel_market',
]

df = pd.read_csv(DATA_PATH, usecols=cols)
df.head()
```

```python
df.shape, df.isna().sum()
```

```python
df.nunique().sort_values(ascending=False)
```

```python
df.to_csv(OUT_PATH, index=False)
OUT_PATH
```

## Optional: unique ID combinations

Если аналитикам не нужны все строки, можно передать только уникальные комбинации.

```python
unique_ids = df.drop_duplicates().reset_index(drop=True)
unique_ids.to_csv('test_location_ids_unique.csv', index=False)
unique_ids.head()
```
